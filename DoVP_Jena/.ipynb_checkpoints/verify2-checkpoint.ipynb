```python
%matplotlib ipympl
from mpl_toolkits.mplot3d import axes3d
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.font_manager import FontProperties
from scipy.special import genlaguerre    #genlaguerre(n, alpha)
import matplotlib as mpl
from mpl_toolkits.axes_grid1 import make_axes_locatable
from matplotlib.pyplot import MultipleLocator

from matplotlib.lines import Line2D
import ipywidgets as widgets
from matplotlib.colors import LinearSegmentedColormap
import matplotlib.ticker as ticker
```

```python
# Define the function
def lambda_1s(k, a_B):
    numerator = np.sqrt(2 * np.pi) * a_B
    denominator = (1 + (a_B / 2)**2 * k**2)**(1.5)
    return numerator / denominator

# Physical constants given by user
a_B = 1.0  # Bohr radius in nm

# Generate k values (Wavevector magnitude in inverse nanometers)
k_vals = np.linspace(0, 10, 500)
lambda_vals = lambda_1s(k_vals, a_B)

# Normalize the result so the highest point is 1
lambda_vals_normalized = lambda_vals / np.max(lambda_vals)

# Plotting
plt.figure(figsize=(6.5, 4.5), dpi=150)
plt.plot(k_vals, lambda_vals_normalized, color='deeppink', linewidth=2.5, 
         label=r'Normalized $\Lambda_{1s,\mathbf{0}}(vc\boldsymbol{k})$')

# Titles and Axis Labels with explicit units
plt.title(r'Normalized 1D Profile of $\Lambda_{1s,\mathbf{0}}(vc\boldsymbol{k})$ ($a_B = 1\text{ nm}$)', fontsize=12, pad=12)
plt.xlabel(r'$|\boldsymbol{k}|$ ($\text{nm}^{-1}$)', fontsize=11)
plt.ylabel('Normalized Amplitude (a.u.)', fontsize=11)

# Polish plot appearance
plt.xlim(0, 10)
plt.ylim(0, 1.05)  # Peak is at 1.0, adding 0.05 headroom for clean viewing
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(frameon=True, facecolor='white', edgecolor='none')

plt.tight_layout()
plt.savefig('normalized_profile.png')
```
