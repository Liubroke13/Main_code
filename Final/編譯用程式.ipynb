```python
project_dir = r"H:\Liu\code\Main_code\Final"

# 要清除的副檔名
extensions = [
    "*.aux", "*.log", "*.toc", "*.out",
    "*.lof", "*.lot",
    "*.fls", "*.fdb_latexmk",
    "*.bbl", "*.blg", "*.bcf", "*.run.xml",
    "*.synctex.gz"
]

os.chdir(project_dir)

deleted = 0

for ext in extensions:
    for file in glob.glob(ext):
        try:
            os.remove(file)
            deleted += 1
        except Exception as e:
            print(f"⚠️ 無法刪除 {file}: {e}")

print(f"🧹 清理完成，共刪除 {deleted} 個 LaTeX 暫存檔")
```

```python
project_dir = r"H:\Liu\code\Main_code\Final"
tex_file = "RY-Master-Dissertation-v1.tex"
basename = tex_file.replace(".tex", "")

os.chdir(project_dir)
def run(cmd):
    # 加入 stdout/stderr 的編碼設定
    r = subprocess.run(
        cmd,
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE,
        text=True,
        encoding='utf-8',       # 明確指定使用 UTF-8
        errors='replace'       # 遇到無法解碼的字元用 "?" 取代，而不是直接崩潰
    )
    
    if r.returncode != 0:
        print("===== 錯誤發生！詳細訊息如下 =====")
        print(r.stdout)  # LaTeX 錯誤通常記錄在 stdout 中
        print(r.stderr)
        raise RuntimeError("XeLaTeX 編譯失敗")

# 1st pass
run(["xelatex", "-interaction=nonstopmode", "-halt-on-error", tex_file])


# bibliography（如果有）
run(["biber", basename])

# 2nd & 3rd pass
run(["xelatex", "-interaction=nonstopmode", "-halt-on-error", tex_file])
run(["xelatex", "-interaction=nonstopmode", "-halt-on-error", tex_file])

print("🎉 XeLaTeX 編譯完成，請查看 SumatraPDF（會自動刷新）")
```
