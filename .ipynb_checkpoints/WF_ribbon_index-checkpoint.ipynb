```python
import numpy as np
import os
Na = 19
base_path=r'G:\Chin Jiu Huang\MoS2\woSOC_prinstine\Na19\ribbon_bin_file'
```

```python
S_total = 6*Na
S_nano_b_l = np.arange(1,12+1)
S_nano_b_r = np.arange(int((Na+1)/2*6-11),int((Na+1)/2*6+1))
S_nano_t_l = np.arange(int((Na+1)/2*6+1),int((Na+1)/2*6+12+1))
S_nano_t_r = np.arange(Na*6-11,Na*6+1)
W_nano_t_l = np.arange(S_total+1,S_total+10+1)
W_nano_t_r = np.arange(int(S_total+(Na+1)/2*5-9),int(S_total+(Na+1)/2*5+1))
W_nano_b_l = np.arange(int(S_total+(Na+1)/2*5+1),int(S_total+(Na+1)/2*5+10+1))
W_nano_b_r = np.arange(S_total+(Na*5)-9,S_total+(Na*5)+1)
print(S_nano_b_l)
print(S_nano_b_r)
print(S_nano_t_l)
print(S_nano_t_r)
print(W_nano_t_l)
print(W_nano_t_r)
print(W_nano_b_l)
print(W_nano_b_r)
```

```python
Na_origin = 11
S_total_origin = 6*Na_origin
S_nano_b_l_origin = np.arange(1,12+1)
S_nano_b_r_origin = np.arange(int((Na_origin+1)/2*6-11),int((Na_origin+1)/2*6+1))
S_nano_t_l_origin = np.arange(int((Na_origin+1)/2*6+1),int((Na_origin+1)/2*6+12+1))
S_nano_t_r_origin = np.arange(Na_origin*6-11,Na_origin*6+1)
W_nano_t_l_origin = np.arange(S_total_origin+1,S_total_origin+10+1)
W_nano_t_r_origin = np.arange(int(S_total_origin+(Na_origin+1)/2*5-9),int(S_total_origin+(Na_origin+1)/2*5+1))
W_nano_b_l_origin = np.arange(int(S_total_origin+(Na_origin+1)/2*5+1),int(S_total_origin+(Na_origin+1)/2*5+10+1))
W_nano_b_r_origin = np.arange(S_total_origin+(Na_origin*5)-9,S_total_origin+(Na_origin*5)+1)
print(S_nano_b_l_origin)
print(S_nano_b_r_origin)
print(S_nano_t_l_origin)
print(S_nano_t_r_origin)
print(W_nano_t_l_origin)
print(W_nano_t_r_origin)
print(W_nano_b_l_origin)
print(W_nano_b_r_origin)
```

```python
dir_name = f'{S_nano_b_l_origin[0]}-{S_nano_b_l_origin[-1]}'
# 設置目錄路徑
directory = os.path.join(base_path,dir_name)
# 收集所有以'.bin'結尾的文件
bin_files = [filename for filename in os.listdir(directory) if filename.endswith('.bin')]
def custom_sort(filename):
    parts = filename.split('_')
    return int(parts[0])

# 使用自訂的排序函數來排序檔案
sorted_files = sorted(bin_files, key=custom_sort)
print(sorted_files)
new_number = S_nano_b_l[0]
real_number = new_number
imag_number = new_number
# 遍歷目錄中的文件
for filename in sorted_files:
    if filename.endswith('_WF_imag.bin'):
        print(f'Processing file: {filename}')  # 添加這行語句來調試
        # 提取原始文件名中的數字部分
        parts = filename.split('_')
        #print(parts[0])
        new_filename = f'{imag_number}_WF_imag.bin'
        old_path = os.path.join(directory, filename)
        new_path = os.path.join(directory, new_filename)   
        #print(new_number)
        imag_number = imag_number+1
        # 重命名文件
        os.rename(old_path, new_path)
        print(f'Renamed: {filename} -> {new_filename}')
print('------------------------')
for filename in sorted_files:
    if filename.endswith('_WF_real.bin'):
        print(f'Processing file: {filename}')  # 添加這行語句來調試
        # 提取原始文件名中的數字部分
        parts = filename.split('_')
        #print(parts[0])
        new_filename = f'{real_number}_WF_real.bin'
        old_path = os.path.join(directory, filename)
        new_path = os.path.join(directory, new_filename)   
        #print(new_number)
        real_number = real_number+1
        # 重命名文件
        os.rename(old_path, new_path)
        print(f'Renamed: {filename} -> {new_filename}')
```
