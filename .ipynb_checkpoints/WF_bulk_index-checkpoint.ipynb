```python
import numpy as np
import os
Na = 15
base_path=r'G:\Wen Teng Yang\20231227 MoS2\Na15\trans\bulk_bin_file'
```

```python
S_total = Na*6
bulk_S_1 = np.arange(13,int((Na+1)/2*6)-11)
bulk_S_2 = np.arange(int((Na+1)/2*6+12+1),Na*6-11)
bulk_W_1 = np.arange(S_total+10+1,S_total+int((Na+1)/2*5)-9)
bulk_W_2 = np.arange(S_total+int((Na+1)/2*5)+11,S_total+(Na*5)-9)
print(bulk_S_1)
print(bulk_S_2)
print(bulk_W_1)
print(bulk_W_2)

directory_S = os.path.join(base_path,'S')
directory_W = os.path.join(base_path,'W')
```

```python
#for S atom bin file
index = bulk_S_1[0]
prefix_imag = "_WF_imag.bin"
prefix_real = "_WF_real.bin"

# 設定開始的索引和前綴
for atom_idx in range(5,Na-3,2):
    prefix = f"atom{atom_idx}_"
    start_imag_index = index
    start_real_index = index
    # 列出目前目錄中的所有符合條件的檔案
    for filename in os.listdir(directory_S):
        if filename.startswith(prefix) and filename.endswith("imag.bin"):
            # 建立新的檔案名稱
            new_name = f"{start_imag_index}{prefix_imag}"
            print(f'Renamed: {filename} -> {new_name}')

            # 建立舊檔案和新檔案的絕對路徑
            old_path = os.path.join(directory_S, filename)
            new_path = os.path.join(directory_S, new_name)

            # 重新命名檔案
            os.rename(old_path, new_path)

            # 更新索引
            start_imag_index = start_imag_index + 1


    print('----------------------')   

    # 列出目前目錄中的所有符合條件的檔案
    for filename in os.listdir(directory_S):
        if filename.startswith(prefix) and filename.endswith("real.bin"):
            # 建立新的檔案名稱
            new_name = f"{start_real_index}{prefix_real}"
            print(f'Renamed: {filename} -> {new_name}')

            # 建立舊檔案和新檔案的絕對路徑
            old_path = os.path.join(directory_S, filename)
            new_path = os.path.join(directory_S, new_name)
            # 重新命名檔案
            os.rename(old_path, new_path)

            # 更新索引
            start_real_index = start_real_index+1 
    print('----------------------') 
    index = start_real_index
```

```python
#for W atom bin file

# 設定前綴和檔案類型
file_type_imag = "_WF_imag.bin"
file_type_real = "_WF_real.bin"
prefix_imag = "_WF_imag.bin"
prefix_real = "_WF_real.bin"

index = bulk_W_1[0]
for atom_idx in range(5,Na-3,2):
    prefix = f"atom{atom_idx}_bulk"

    start_imag_index_W = index
    start_real_index_W = index

    # 列出目前目錄中的所有符合條件的檔案
    files_imag = [filename for filename in os.listdir(directory_W) if filename.startswith(prefix) and filename.endswith(file_type_imag)]

    # 定義一個自訂的排序函數，以數字部分來排序
    def custom_sort(filename):
        parts = filename[len(prefix):].split('_')
        return int(parts[0])

    # 使用自訂的排序函數來排序檔案
    sorted_files_imag = sorted(files_imag, key=custom_sort)

    # 打印排序後的檔案名稱
    for filename in sorted_files_imag:

        new_name = f"{start_imag_index_W}{prefix_imag}"
        print(f'Renamed: {filename} -> {new_name}')

        # 建立舊檔案和新檔案的絕對路徑
        old_path = os.path.join(directory_W, filename)
        new_path = os.path.join(directory_W, new_name)
        # 重新命名檔案
        os.rename(old_path, new_path) 
        start_imag_index_W =start_imag_index_W +1
    print('----------------')    
    # 列出目前目錄中的所有符合條件的檔案
    files_real = [filename for filename in os.listdir(directory_W) if filename.startswith(prefix) and filename.endswith(file_type_real)]

    # 定義一個自訂的排序函數，以數字部分來排序
    def custom_sort(filename):
        parts = filename[len(prefix):].split('_')
        return int(parts[0])

    # 使用自訂的排序函數來排序檔案
    sorted_files_real = sorted(files_real, key=custom_sort)

    # 打印排序後的檔案名稱
    for filename in sorted_files_real:

        new_name = f"{start_real_index_W}{prefix_real}"
        print(f'Renamed: {filename} -> {new_name}')

        # 建立舊檔案和新檔案的絕對路徑
        old_path = os.path.join(directory_W, filename)
        new_path = os.path.join(directory_W, new_name)
        # 重新命名檔案
        os.rename(old_path, new_path)    

        start_real_index_W = start_real_index_W+1
    print('----------------')  
    index = start_real_index_W
```
